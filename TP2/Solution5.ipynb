```python
#step 1
prices = {"apple": 0.50, "banana": 0.25, "mango": 1.00} 
print("Price of mango:", prices["mango"])
```

```python
#step 2
prices["orange"] = 0.75  #prices.update({"orange": 0.75, "grape": 0.90})
prices["apple"] = 0.60 
del prices["banana"] 
print("Prices:", prices)
```

```python
#step 3
print("Price of grape:", prices.get("grape", "not sold"))
```

```python
#step 4
for fruit, price in prices.items(): 
    print(f"{fruit:<8} ${price:.2f}")
```

```python
#step 5
basket = {"apple": 4, "mango": 2, "orange": 3} 
total = 0 
for fruit, qty in basket.items(): total += prices[fruit] * qty 
print(f"Total bill: ${total:.2f}")
```
