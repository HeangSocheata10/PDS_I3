A list can grow and shrink. Build a shopping list and change it step by step. Print the list after each step to see what happened.

Step 1. Create the list with three items and print it.

```python
student = ("Dara",20, 3.45)
print("Name: ", student[0])
```

Step2.Unpack the tuple: put its three values into three variables in one line.



```python
name, age, gpa = student
print(name , "is ", age, " years old, GPA ", gpa)

'''
it is the same as :
name = student[0]
age = student[1]
gpa = student[2]
'''
```

Step 3. Make a list of four student tuples and print each one with a for loop that unpacks the tuple.

```python
students = [("Dara",20,3.45), ("Sophea",21,3.92),
            ("Vibol",19,3.10),("Malis",22,3.78)]
for name, age ,gpa in students:
    print(name,age,gpa)
```

Step 4. Sort the students by GPA, best first. key=lambda s: s[2] tells sorted() to compare the item at index 2 (the GPA). The best student is the first tuple of the result.

```python
ranking = sorted(students, key = lambda s: s[2], reverse = True)  # sorted(iterable, key=None, reverse=False)
print(ranking)
print (ranking[0])
print("Best student : ", ranking[0][0])
```

Step 5. Try to change the GPA inside the tuple. Python refuses; try/except catches the error so the program can continue.

```python
try: 
    student[2] = 4.0 
except TypeError  as error: 
    print("Error: ", error)
```
